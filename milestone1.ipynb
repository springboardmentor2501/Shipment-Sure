```python
#1. Dataset Overview
import pandas as pd

df = pd.read_csv("/content/smart_logistics_dataset.csv")

df.head()
df.shape
df.info()
df.describe()
```

```python
#3. Univariate Analysis
#Numerical Features
import matplotlib.pyplot as plt
import seaborn as sns

num_cols = ['Inventory_Level','Temperature','Humidity','Waiting_Time',
            'User_Transaction_Amount','User_Purchase_Frequency',
            'Asset_Utilization','Demand_Forecast']

for col in num_cols:
    sns.histplot(df[col], kde=True)
    plt.title(f"Distribution of {col}")
    plt.show()
```

```python
#Categorical Features
cat_cols = ['Shipment_Status','Traffic_Status','Logistics_Delay_Reason']

for col in cat_cols:
    sns.countplot(x=df[col])
    plt.title(f"Count of {col}")
    plt.xticks(rotation=45)
    plt.show()
```

```python
#4. Bivariate Analysis With Target: Logistics_Delay
sns.boxplot(x='Logistics_Delay', y='Waiting_Time', data=df)
plt.show()

sns.boxplot(x='Logistics_Delay', y='Temperature', data=df)
plt.show()

sns.countplot(x='Traffic_Status', hue='Logistics_Delay', data=df)
plt.xticks(rotation=45)
plt.show()
```

```python
#5. Feature Relationships
plt.figure(figsize=(10,6))
sns.heatmap(df[num_cols].corr(), annot=True, cmap='coolwarm')
plt.title("Feature Correlation Matrix")
plt.show()
```

```python
#6. Class Imbalance Check
sns.countplot(x='Logistics_Delay', data=df)
plt.title("Class Distribution of Logistics_Delay")
plt.show()

df['Logistics_Delay'].value_counts(normalize=True)
```
