```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("seaborn-v0_8")
sns.set_palette("Set2")

# Load Excel file
df = pd.read_excel("shipment_dataset_10000 -1.xlsx")
```

```python
#Dataset Overview
```

```python
print("Dataset Shape:", df.shape)
df.head()
```

```python
#Target Variable Analysis
```

```python
# Basic info: data types, missing values
df.info()

# Descriptive statistics for numerical features
df.describe()

# Check for nulls in the dataset
df.isnull().sum()

# Check shape of the dataset
print(f"Dataset has {df.shape[0]} rows and {df.shape[1]} columns")
```

```python
#Univariate Analysis
```

```python
num_cols = df.select_dtypes(include=['int64', 'float64']).columns

for col in num_cols:
    plt.figure(figsize=(6,4))
    sns.histplot(df[col], kde=True, bins=30)
    plt.title(f'Distribution of {col}')
    plt.show()
```

```python
#categorical feature
```

```python
cat_cols = df.select_dtypes(include=['object', 'category']).columns

for col in cat_cols:
    plt.figure(figsize=(6,4))
    sns.countplot(data=df, x=col, order=df[col].value_counts().index)
    plt.title(f'Count plot of {col}')
    plt.xticks(rotation=45)
    plt.show()
```

```python
#Bivariate Analysis
```

```python
plt.figure(figsize=(12,6))
sns.boxplot(data=df, x='on_time_delivery', y='supplier_lead_time')
plt.title('Supplier Lead Time vs On-Time Delivery')
plt.show()

plt.figure(figsize=(12,6))
sns.boxplot(data=df, x='on_time_delivery', y='shipping_distance_km')
plt.title('Shipping Distance vs On-Time Delivery')
plt.show()
```

```python
#Categorical vs Target
```

```python
for col in cat_cols:
    if col != 'on_time_delivery':
        plt.figure(figsize=(8,4))
        sns.countplot(data=df, x=col, hue='on_time_delivery')
        plt.title(f'{col} vs On-Time Delivery')
        plt.xticks(rotation=45)
        plt.show()
```

```python
#Correlation Analysis
```

```python
plt.figure(figsize=(12,8))
corr = df[num_cols].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap of Numerical Features')
plt.show()
```

```python
#Target Variable Analysis
```

```python
# Class distribution
sns.countplot(data=df, x='on_time_delivery')
plt.title('Distribution of On-Time Delivery')
plt.show()

# Percentage distribution
target_counts = df['on_time_delivery'].value_counts(normalize=True)*100
print("Percentage distribution of target variable:\n", target_counts)
```

```python
date_cols = ['order_date', 'promised_delivery_date', 'actual_delivery_date']
df[date_cols].info()
```

```python
df['delivery_delay_days'] = (
    df['actual_delivery_date'] - df['promised_delivery_date']
).dt.days
```

```python
sns.boxplot(x='on_time_delivery', y='delivery_delay_days', data=df)
plt.title("Delivery Delay vs On-Time Delivery")
plt.show()
```

```python
#Outlier Detection(Numerical)
```

```python
for col in [
    'supplier_lead_time',
    'shipping_distance_km',
    'order_quantity',
    'unit_price',
    'total_order_value'
]:
    plt.figure(figsize=(6,3))
    sns.boxplot(x=df[col])
    plt.title(f'Outlier Detection: {col}')
    plt.show()
```
